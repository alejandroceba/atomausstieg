```python
import pandas as pd
import matplotlib.pyplot as plt

f = "../data/bruttostromerzeugung_deutschland_1990_2025.csv"
erzeugungtwh = pd.read_csv(f, sep=";", index_col=0, parse_dates=True)
erzeugungtwh.columns = [c.replace(" (TWh)", "") for c in erzeugungtwh.columns]

energietraeger = erzeugungtwh.drop(columns="Gesamterzeugung")

fig, ax = plt.subplots(figsize=(12, 7))

ax.stackplot(
    erzeugungtwh.index.year,
    energietraeger.T,
    labels=energietraeger.columns,
    alpha=0.9
)

ax.set_ylabel("Bruttostromerzeugung in TWh")
ax.legend(loc="upper left", bbox_to_anchor=(1, 1))
ax.grid(axis="y", alpha=0.3)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

f = "../data/bruttostromerzeugung_deutschland_1990_2025.csv"
erzeugungtwh = pd.read_csv(f, sep=";", index_col=0, parse_dates=True)
erzeugungtwh.columns = [c.replace(" (TWh)", "") for c in erzeugungtwh.columns]

erzeugungprz = erzeugungtwh.copy()
for col in erzeugungprz.columns:
    erzeugungprz[col] = erzeugungprz[col] * 100 / erzeugungprz.Gesamterzeugung

energietraeger = erzeugungprz.drop(columns="Gesamterzeugung")

fig, ax = plt.subplots(figsize=(12, 7))

ax.stackplot(
    erzeugungprz.index.year,
    energietraeger.T,
    labels=energietraeger.columns,
    alpha=0.9
)

ax.set_ylabel("Bruttostromerzeugung %")
ax.legend(loc="upper left", bbox_to_anchor=(1, 1))
ax.grid(axis="y", alpha=0.3)
```

```python
import matplotlib.pyplot as plt

f = "../data/statistischer-bericht-energiepreisentwicklung-5619001.xlsx"
kosten = pd.read_excel(f, sheet_name="61241-15", skiprows=3, usecols=[0,3])

kosten = kosten.iloc[1:-1].copy()
kosten.columns = ["halbjahr", "euro_pro_kwh"]
kosten["jahr"] = kosten["halbjahr"].str.extract(r"(\d{4})").astype(int)
kosten["halbjahr_nummer"] = kosten["halbjahr"].str.extract(r"(1|2)\. HJ").astype(int)
kosten["x"] = kosten["jahr"] + (kosten["halbjahr_nummer"] - 1) / 2
kosten["ct_pro_kwh"] = kosten["euro_pro_kwh"] * 100

fig, ax = plt.subplots(figsize=(12, 7))
ax.plot(kosten["x"], kosten["ct_pro_kwh"], marker="o", linewidth=2)

ax.set_xlabel("Jahr")
ax.set_ylabel("ct/kWh")
ax.set_xticks(range(kosten["jahr"].min(), kosten["jahr"].max() + 1, 2))
ax.grid(axis="y", alpha=0.3)
plt.show()
```
